```python
# Task 1.1
#MongoDB setup for ParcelDB and Parcels collection

#Imports
import json
import pymongo

#Main Program
client = MongoClient('127.0.0.1', 27017)
db = client["ParcelDB"]
parcels = db["Parcels"]
```

```python
# Task 1.2


#Functions & Procedures
def loadparcels():
    """Read parcel records from TASK1_2.JSON and insert them into MongoDB."""
    with open("TASK1_2.JSON", "r") as file:
        parcel_data = json.load(file)
    #file automatically closed

    #clear existing documents to repeated testing does not create duplicates
    parcels.delete_many({})

    if len(parcel_data) > 0: #validation
        parcels.insert_many(parcel_data)
        print("Parcel records loaded successfully.")
    else:
        print("No parcel records found in JSON file.")
```

```python
# Task 1.3

def displayParcel(parcelID):
    """Search for and display one parcel based on parcelID."""
    parcel = parcels.find_one({'parcelID': parcelID})

    if parcel is not None:
        print("---Parcel Found---")
        print("Parcel ID: ", parcel["parcelID"])
        print("Recipient: ", parcel["recipient"])
        print("Priority: ", parcel["priority"])
        print("Status: ", parcel["status"])
    else:
        print("---Parcel Not Found---")
```

```python
# Task 1.4

def priorityParcels(priorityType):
    """Display parcel ID and recipient for all parcels of the given priority type"""
    results = parcels.find({'priority': priorityType})
    count = 0
    print(f"{'Parcel ID':<12}{'Recipient'<20}")
    for parcel in results:
        print(f"{parcel['parcelID']:12}{parcel['recipient']:<20}")
        count +=1
    print("Total no. of parcels: ", count)
```

```python
# Main
if __name == "__main__":
    loadparcels()
    print()

    displayParcel("P1001")
    print()

    priorityParcels("Standard") #Task 1.5
    print()
```
