```python
#Task 1.1

#Imports
import sqlite3

#Functions & Procedures
def init_db():

    drop_queries = ['DROP TABLE IF EXISTS Student_Test', 'DROP TABLE IF EXISTS Test', 'DROP TABLE IF EXISTS Class', 'DROP TABLE IF EXISTS Student']
    student_query = 'CREATE TABLE Student (StudentID INTEGER PRIMARY KEY NOT NULL, GivenName TEXT, FamilyName TEXT, ClassName TEXT)'
    test_query = 'CREATE TABLE Test (TestID TEXT PRIMARY KEY NOT NULL, MaxMarks INTEGER)'
    class_query = 'CREATE TABLE Class (ClassName TEXT PRIMARY KEY NOT NULL, YearGroup INTEGER)'
    student_test_query = 'CREATE TABLE Student_Test (StudentTestID INTEGER PRIMARY KEY NOT NULL, StudentID INTEGER, TestID TEXT, Mark INTEGER NOT NULL, Date TEXT NOT NULL, FOREIGN KEY (StudentID) REFERENCES Student(StudentID), FOREIGN KEY (TestID) REFERENCES Test(TestID))'
    main_queries = [student_query, test_query, class_query, student_test_query]

    connection = sqlite3.connect("TESTDATABASE.db") #Connects program to database, creates database if not exists
    cursor = connection.cursor() #Cursor object used to perform operations on database

    #Drop tables
    for query in drop_queries:
        cursor.execute(query)
    #Create tables
    for query in main_queries:
        cursor.execute(query)

    connection.commit() #Saves changes
    connection.close() #Terminates connection to database

    print("Database creation successful")
#Main Program
init_db()
#Testing
```
