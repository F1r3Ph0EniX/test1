```python
#Task 2.1

def task2_1(filename):

    integer_list = []

    #Obtaining values from file
    with open(filename, "r") as file:
        integers = file.readlines()
    #File is automatically closed

    #converting values from string to integer
    for integer in integers:
        integer = int(integer)
        integer_list.append(integer)

    return integer_list

#Testing
result = task2_1('TEN.txt')
print(result)
print(len(result))
```

```python
#Task 2.2

def task2_2(list_of_integers):

    for i in range(len(list_of_integers)-1):
        i += 1
        current = list_of_integers[i]
        j = i-1
        while j >= 0 and list_of_integers[j] > current: #Comparing if previous numbers are smaller than current number
            list_of_integers[j + 1] = list_of_integers[j]
            j -= 1
        list_of_integers[j+1] = current

    return list_of_integers

#Testing
print(task2_2(task2_1('TEN.txt')))
        
```

```python
#Task 2.3

def task2_3(list_of_integers):

    def partition(integerList, low, high):

        pivot = integerList[low] #first element as pivot
        i = low
        j = low + 1
        while j <= high:
            #Case if pivot is greater than current value
            if integerList[j] <= pivot:
                i += 1
                integerList[i], integerList[j] = integerList[j], integerList[i] #Swapping values
            j += 1

        integerList[i], integerList[low] = integerList[low], integerList[i]

        return i

    def quicksort(integerList, low, high):
        
        if low < high:
            pivot_index = partition(integerList, low, high)
            quicksort(integerList, low, pivot_index - 1)
            quicksort(integerList, pivot_index +1, high)

    quicksort(list_of_integers, 0, len(list_of_integers)-1)
    return list_of_integers

#Testing
print(task2_3(task2_1('TEN.txt')))
```

```python
#Task 2.4

#Imports
import timeit

ten_integers = task2_1('TEN.txt')
hundred_integers = task2_1('HUNDRED.txt')
thousand_integers = task2_1('THOUSAND.txt')

#Testing time taken for Insertion Sort and Quicksort to execute once
print("Time taken for Insertion Sort to sort 10 integers: ", timeit.timeit(lambda: task2_2(ten_integers), number=1))
print("Time taken for Quicksort to sort 10 integers: ", timeit.timeit(lambda: task2_3(ten_integers), number=1))

print("Time taken for Insertion Sort to sort 100 integers: ", timeit.timeit(lambda: task2_2(hundred_integers), number=1))
print("Time taken for Quicksort to sort 100 integers: ", timeit.timeit(lambda: task2_3(hundred_integers), number=1))

print("Time taken for Insertion Sort to sort 1000 integers: ", timeit.timeit(lambda: task2_2(thousand_integers), number=1))
print("Time taken for Quicksort to sort 1000 integers: ", timeit.timeit(lambda: task2_3(thousand_integers), number=1))
```
