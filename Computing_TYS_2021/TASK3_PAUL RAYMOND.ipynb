```python
#Task 3.1

class Node:

    def __init__(self, data):
        self.data = data
        self.pointer = None

    #Get & Set Methods

    def get_data(self):
        return self.data

    def set_data(self, newdata):
        self.data = newdata

    def get_pointer(self):
        return self.pointer

    def set_pointer(self, newpointer):
        self.pointer = newpointer

class LinkedList:

    def __init__(self):
        self.head = None

    #Get & Set Methods

    def get_head(self):
        return self.head

    def set_head(self, newhead):
        self.head = newhead

    #Insertion method

    def insert(self, integer_value):
        '''INSERT AT HEAD??'''
        newNode = Node(integer_value)

        #case if Linked List is empty
        if self.get_head() is None:
            self.set_head(newNode)
        #Case if otherwise
        else:
            #Traverse to end of list
            current = self.get_head()
            while current.get_pointer() is not None:
                current = current.get_pointer()

            current.set_pointer(newNode)

    def delete(self, integer_value):

        #Case if Linked List is empty
        if self.get_head() is None: 
            return None

        #Case if integer is in first (head) node
        if self.get_head().get_data() == integer_value:
            self.set_head(self.get_head().get_pointer())
            return 

        #Search for integer in rest of list
        previous = self.get_head()
        current = previous.get_pointer()

        while current is not None:
            if current.get_data() == integer_value: #Case if found
                #rerouting pointer to point to next node
                previous.set_pointer(current.get_pointer())
                return

            #Increment pointers
            previous = current
            current = current.get_pointer()
                
        return None #Case if integer not found after Linked List has been fully traversed

    def search(self, integer_value):

        #Case if Linked List is empty
        if self.get_head() is None: 
            return False

        current = previous.get_pointer()
        while current is not None:
            if current.get_data() == integer_value: #Case if found
                #rerouting pointer to point to next node
                return True
            #Increment pointers
            current = current.get_pointer()

        return False

    def count(self):
        counter = 0
        previous = self.get_head()
        current = previous.get_pointer()
        while current is not None:
            counter += 1
            previous = current
            current = current.get_pointer()

        return counter

    def to_String(self):
        string_list = []
        previous = self.get_head()
        current = previous.get_pointer()

        string_list.append(previous.get_data())

        while current is not None:
            string_list.append(current.get_data())

            previous = current
            current = current.get_pointer()

        return string_list

#Testing
LL = LinkedList()
LL.insert(11)
LL.insert(2)
LL.insert(7)
LL.insert(4)
LL.to_String()
            
```

```python
#Task 3.2

class SortedLinkedList(LinkedList):
    def __init__(self):
        #Calling parent class to initialise
        super().__init__()

    def insert(self, integer_value):
        newNode = Node(integer_value)
        
        #Case if Linked List is empty
        if self.get_head() is None or integer_value < self.get_head().get_data():
            newNode.set_pointer(self.get_head())
            self.set_head(newNode)
            return

        #Case if otherwise
        previous = self.get_head()
        current = previous.get_pointer()
        
        while current is not None and current.get_data() < integer_value:
            previous = current
            current = current.get_pointer()

        #Insert node
        newNode.set_pointer(current)
        previous.set_pointer(newNode)

#Testing
SLL = SortedLinkedList()
SLL.insert(18)
SLL.insert(5)
SLL.insert(27)
print(SLL.to_String())
SLL.delete(5)
SLL.insert(247)
SLL.count()
```
