```python
#Task 1.1

#Function
def task1_1(input_value):
    #Initialising variables
    valid = True

    #Validation
    if type(input_value) == int:
        input_value = str(input_value)
        length = len(input_value)
        if length != 14:
                valid = False
        
    elif type(input_value) == str:
            if input_value.isdigit() == True:
                valid = True
            length = len(input_value)

            if length != 14:
                valid = False
        
    else:
        valid = False

    if not valid:
        return -1

    sum_value = 0
    number = input_value[::-1]
    #print(number) #Debugging

    for i in range(len(number)):
        
        #Case if odd
        if (i+1) % 2 == 1:
            digit = int(number[i])*2
            digit = str(digit)
            odd_sum = 0
            for character in digit:
                odd_sum += int(character)

            digit = odd_sum

            #print("Odd position: ",number[i]) #Debugging
            #print("Digit: ",digit) #Debugging
            
        #Case if even
        else:
            digit = int(number[i])
            #print("Even position: ",digit) #Debugging
        
        sum_value += digit
        #print("Current sum value: ", sum_value, " Digit added: ", digit) #debugging
    counter = 0
    check_digit = 0
    while counter <= 9:
        if (sum_value + counter) % 10 == 0:
            check_digit = counter
            break
        counter += 1
        
    #print("Sum value: ",sum_value, "Check digit: ", check_digit) #debugging
    return check_digit
    
    

#Main Program

#Testing
task1_1(14576567654934)
#task1_1(1234)
#task1_1(True)
```

```python
#Task 1.2

print(task1_1("12132435465768") == 4)
print(task1_1(47362847592731) == 1)
print(task1_1("hello world") == -1)
print(task1_1(1234) == -1)
```

```python
#Task 1.3

def task1_3(input_value):

    #Validation
    valid = True
    if type(input_value) == int:
        input_value = str(input_value)
        length = len(input_value)
        if length != 15:
                valid = False
        
    elif type(input_value) == str:
            if input_value.isdigit() == True:
                valid = True
            length = len(input_value)

            if length != 15:
                valid = False
        
    else:
        valid = False

    if not valid:
        return False

    number = input_value[0:-1]
    check_digit = input_value[-1]
    #print(number, check_digit) #Debugging

    calculated_check_digit = task1_1(number)
    #print("Calculated check digit: ", calculated_check_digit) #debugging
    #print("Given check digit: ", check_digit) #debugging

    return int(check_digit) == int(calculated_check_digit)

#Testing
print(task1_3("121324354657684"))
print(task1_3(473628475927311))
print(not task1_3("Hehe"))
print(not task1_3(1234))
```
