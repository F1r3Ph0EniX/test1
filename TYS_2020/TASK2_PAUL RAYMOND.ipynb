```python
#Task 2.3

def task2_3(filename_in, filename_out):

    integers_list = []
    #Reading integers from input file
    with open(filename_in, "r") as file:
        integers = file.readlines()
        for integer in integers:
            integer = int(integer.strip())
            integers_list.append(integer)
        #print(integers_list) #Debugging
    #file automatically closed

    sorted_list = task2_2(integers_list)
    #print(sorted_list) #Debugging

    #Writing integers into output file
    with open(filename_out, "w") as file:
        for i in range(len(sorted_list)):
            integer = str(sorted_list[i])
            file.write(integer + '\n')
    #file automatically closed


    

#Testing function
task2_3("randomnumbers_Paul Raymond.txt", "sortednumbers_Paul Raymond.txt")
```
