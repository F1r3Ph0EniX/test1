```python
#Task 3.1

#Functions
def task3_1(quantity_of_data):
    '''Function Description'''

    #initialising variables
    invalid = False
    digits = ""
    units = ""

    #extracting and filtering data
    for character in quantity_of_data:
        if character.isdigit() == True:
            digits += character
        elif character.isalpha() == True:
            units += character
        else:
            invalid = True

    #print(digits) #debugging
    #print(units) #debugging

    #Checking if units are invalid or no digits entered
    if units != "KB" and units != "MB" and units != "GB" and units != "TB" or digits == "":
        invalid = True

    #Case if input is invalid
    if invalid == True:
        print("invalid data")
        return "invalid data"
    else:
        #multiplying digits by respective weights
        if units == "KB":
            digits = int(digits)*(10**3)
        elif units == "MB":
            digits = int(digits)*(10**6)
        elif units == "GB":
            digits = int(digits)*(10**9)
        elif units == "TB":
            digits = int(digits)*(10**12)

    #displays and returns digits
    print(digits)
    return digits

#Main Program

#Testing
task3_1("8KB") #normal case
task3_1("1000mB") #error case
task3_1("256GB") #normal case
task3_1("tB") #error case
task3_1("1TB") #normal case
    
```

```python
#Task 3.2

def task3_2(quantity_of_data):
    '''Function Description'''

    #Initialising variables
    valid_units = ["KB","KiB","MB","MiB","GB","GiB","TB","TiB"]
    invalid = False
    digits = ""
    units = ""

    #extracting and filtering data
    for character in quantity_of_data:
        if character.isdigit() == True:
            digits += character
        elif character.isalpha() == True:
            units += character
        else:
            invalid = True

    #print(digits) #debugging
    #print(units) #debugging

    #checking if units are invalid or no digits entered
    if units not in valid_units or digits == "":
        invalid = True

    #case if input is invalid
    if invalid == True:
        print("invalid data")
        return "invalid data"
    else:
        #multiplying digits by respective weights
        if units == "KB":
            digits = int(digits)*(10**3)
        elif units == "KiB":
            digits = int(digits)*(2**10)
        elif units == "MB":
            digits = int(digits)*(10**6)
        elif units == "MiB":
            digits = int(digits)*(2**20)
        elif units == "GB":
            digits = int(digits)*(10**9)
        elif units == "GiB":
            digits = int(digits)*(2**30)
        elif units == "TB":
            digits = int(digits)*(10**12)
        elif units == "TiB":
            digits = int(digits)*(2**40)

    #displays and returns digits
    print(digits)
    return digits

#Testing

#Normal cases
task3_2("2MiB")
task3_2("5KB")
task3_2("15KiB")
task3_2("20MB")
task3_2("21MiB")
task3_2("256GB")
task3_2("254GiB")
task3_2("1TB")
task3_2("10TiB")

#Error cases
task3_2("potato")
task3_2("2miB")
task3_2("TB")
task3_2("")
```

```python
#Task 3.3

def task3_3(quantity_of_data, target_unit):

    #Initialising variables
    valid_units = ["KB","KiB","MB","MiB","GB","GiB","TB","TiB"]
    invalid = False

    if target_unit not in valid_units:
        print("invalid data")
        return "invalid data"

    actual_bytes = int(task3_2(quantity_of_data))

    if target_unit == "KB":
        weight = 10**3
    elif target_unit == "KiB":
        weight = 2**10
    elif target_unit == "MB":
        weight = 10**6
    elif target_unit == "MiB":
        weight = 2**20
    elif target_unit == "GB":
        weight = 10**9
    elif target_unit == "GiB":
        weight = 2**30
    elif target_unit == "TB":
        weight = 10**12
    elif target_unit == "TiB":
        weight = 2**40

    converted_data = actual_bytes/weight

    print(converted_data)
    return converted_data

#Main Program

#Testing

#Normal case
task3_3("512MiB", "GiB")
#Boundary case
task3_3("0GB", "KiB")
#Error case
task3_3("", "")
```
