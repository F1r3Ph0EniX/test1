```python
#Task 1.1

def task1_1(string_value):
    '''Hash function'''
    h = 0 #Initialises hash value
    for i in range(len(string_value)):
        val = 33*ord(string_value[i])
        h = (h + val) % 1024
    return h #Returns hash value of input string

#Testing function
print(task1_1("Hello"))
print(task1_1("Hallo"))
print(task1_1("Hullo"))
```

```python
#Task 1.2

def task1_2(seed, string_value):
    string = seed + string_value #combines 2 strings into a single variable
    #print(string) #Debugging
    return task1_1(string) #passes combined strings into hash function and returns output

#Testing function
print(task1_2("seed-one", "Hello"))
print(task1_2("seed-two", "Hello"))
print(task1_2("seed-three", "Hello"))
```
