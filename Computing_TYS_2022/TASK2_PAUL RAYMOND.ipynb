```python
#Task 2.3

def mergeSortProcedure(ar):

    def mergeSort(ar):

        if len(ar) <= 1:
            return ar

        mid = len(ar) // 2

        left_ar = mergeSort(ar[0:mid])
        right_ar = mergeSort(ar[mid:])

        return merge(left_ar, right_ar)

    def merge(left, right):

        sorted_ar = []

        i = 0
        j = 0

        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                sorted_ar.append(left[i])
                i += 1
            else:
                sorted_ar.append(right[j])
                j += 1

        sorted_ar.extend(left[i:])
        sorted_ar.extend(right[j:])

        return sorted_ar

    return mergeSort(ar)

unsorted_list = task2_1()
choice = input("Which sorting algorithm would you prefer (bubble, merge): ")
while choice != "bubble" and choice != "merge":
    print("Invalid input!")
    choice = input("Which sorting algorithm would you prefer (bubble, merge): ")

if choice == "bubble":
    sorted_list = bubbleSort(unsorted_list)
else:
    sorted_list = mergeSortProcedure(unsorted_list)

print(sorted_list)
```
