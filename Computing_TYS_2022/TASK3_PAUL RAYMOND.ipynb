```python
#Task 3.1

#Imports
import random

#Classes
class Tree:
    def __init__(self, data=None):
        self.left_pointer = None
        self.right_pointer = None
        self.data = data

    def insert(self, data): #Recursive insertion

        if self.data == None:
            self.data = data
        else:
        
            newNode = Tree(data)

            if data > self.data: #new node data greater than current data
                if self.right_pointer == None: #Case if no nodes on right side
                    self.right_pointer = newNode
                else:
                    self.right_pointer.insert(data) #Moves down to node on right to insert
                
            elif data < self.data:
                if self.left_pointer == None:
                    self.left_pointer = newNode
                else:
                    self.left_pointer.insert(data) #Moves down to node on left to insert

    def inorder(self):

        result = []

        if self.left_pointer is not None:
            result.extend(self.left_pointer.inorder())

        result.append(self.data)

        if self.right_pointer is not None:
            result.extend(self.right_pointer.inorder())

        return result

    def postorder(self):

        result = []

        if self.left_pointer is not None:
            result.extend(self.left_pointer.postorder())

        if self.right_pointer is not None:
            result.extend(self.right_pointer.postorder())

        result.append(self.data)

        return result

#Main Program
newTree = Tree()

integer_list = []
for i in range(10):
    number = random.randint(0,999)
    integer_list.append(number)

for integer in integer_list:
    newTree.insert(integer)

print(newTree.inorder()) #Testing
print(newTree.postorder()) #Testing
```
