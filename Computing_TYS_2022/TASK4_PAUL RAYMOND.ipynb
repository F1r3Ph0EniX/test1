```python
#Task 4.1

#Imports
import sqlite3

#Functions
def init_db():

    book_query = "CREATE TABLE Book(BookID INTEGER PRIMARY KEY NOT NULL, Title TEXT NOT NULL, Genre TEXT NOT NULL)"
    member_query = "CREATE TABLE Member(MemberNumber INTEGER PRIMARY KEY NOT NULL, FamilyName TEXT NOT NULL, GivenName TEXT NOT NULL)"
    loan_query = "CREATE TABLE Loan(LoanID INTEGER PRMARY KEY NOT NULL, MemberNumber INTEGER NOT NULL, BookID INTEGER NOT NULL, DateLoaned TEXT NOT NULL, Returned TEXT NOT NULL, FOREIGN KEY(MemberNumber) REFERENCES Member(MemberNumber), FOREIGN KEY(BookID) REFERENCES Book(BookID))"

    queries = [book_query, member_query, loan_query]

    connection = sqlite3.connect("LIBRARY.db") #Connects python program to database

    cursor = connection.cursor() #Cursor object to perform operations in database

    for query in queries:
        cursor.execute(query)

    connection.commit() #Saves changes
    connection.close() #Terminate connection to database

#Main Program
init_db()
```

```python
#Task 4.2

#Imports
import sqlite3

#Functions & Procedures
def insert_data():

    book_data = []
    with open("BOOK.txt", "r") as file:
        raw_data = file.readlines()
    #File automatically closed
    for line in raw_data:
        #print(line) #Debugging
        line = line.strip().split(",")
        bookID = line[0]
        title = line[1]
        genre = line[2]

        book_data.append([bookID, title, genre])

    member_data = []
    with open("MEMBER.txt", "r") as file:
        raw_data = file.readlines()
    #File automatically closed
    for line in raw_data:
        line = line.strip().split(",")
        memberNumber = line[0]
        family = line[1]
        given = line[2]

        member_data.append([memberNumber, family, given])

    loan_data = []
    with open("LOAN.txt", "r") as file:
        raw_data = file.readlines()
    #File automatically closed
    for line in raw_data:
        line = line.strip().split(",")
        loanID = line[0]
        memberNumber = line[1]
        bookID = line[2]
        dateLoaned = line[3]
        returned = line[4]

        loan_data.append([loanID, memberNumber, bookID, dateLoaned, returned])


    book_query = "INSERT INTO Book VALUES (?,?,?)"
    member_query = "INSERT INTO Member VALUES (?,?,?)"
    loan_query = "INSERT INTO Loan VALUES (?,?,?,?,?)"

    connection = sqlite3.connect("LIBRARY.db") #Connects python program to database

    cursor = connection.cursor() #Cursor object to perform operations in database

    cursor.executemany(book_query, book_data)
    cursor.executemany(member_query, member_data)
    cursor.executemany(loan_query, loan_data)

    connection.commit() #Saves changes
    connection.close() #Terminate connection to database

insert_data()
```
