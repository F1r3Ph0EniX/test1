```python
#Task 1.1

def encrypt(character):

    if character != " " and character not in "QWERTYUIOPASDFGHJKLZXCVBNM" and character not in "qwertyuiopasdfghjklzxcvbnm": #Case if character is invalid
        #print("Invalid character") #Debugging
        return -1

    if character == " ": #Case if character is whitespace
        return 33 #ASCII value for '!'

    else:
            
        value = ord(character) #Obtaining ascii value of character
        value += 10 #Incrementing value by 10
        
        if character.isupper() and value > 90: #Case if character is uppercase and has shifted value greater than uppercase characters
            value = value - 91 + 65 #Wraps around value
            
        elif character.islower() and value > 122: #Case if character is lowercase and has shifted value greater than lowercase characters
            value = value - 123 + 97 #Wraps around value
            
        #print(value)
        #print(chr(value)) #Debugging
    return value


#Testing
#encrypt("0")
#encrypt(" ")
#encrypt("A")
#encrypt("a")
#encrypt("Z")
#encrypt("x")
```

```python
#Task 1.3

#Reading data from file
with open("DATATOENCRYPT.txt", "r") as file:
    data = file.readline()
#File automatically closed

#Cleaning data
data = data.strip()

#Encrypting data
encrypted_data = ""

for character in data:
    encrypted_character = encrypt(character)
    if encrypted_character != -1: #Appends only valid character to encrypted message
        encrypted_data += chr(encrypted_character)

#Writing encrypted data to file
with open("ENCRYPTEDMESSAGE.txt", "w") as file:
    file.write(encrypted_data) #Writing encrypted message into file
#File automatically closed
```
